```python
import numpy as np
import moderngl_window as mglw

class App(mglw.WindowConfig):
    gl_version = (3, 3)
    title = "Triangle"
    window_size = (800, 600)

    def __init__(self, **kwargs):
        super().__init__(**kwargs)

        self.prog = self.ctx.program(
            vertex_shader="""
                #version 330
                in vec3 in_pos;
                void main() {
                    gl_Position = vec4(in_pos, 1.0);
                }
            """,
            fragment_shader="""
                #version 330
                out vec4 f_color;
                void main() {
                    f_color = vec4(1.0, 0.6, 0.2, 1.0);
                }
            """,
        )

        # IMPORTANT: float32 binary buffer, nu text
        verts = np.array([
            -0.6, -0.4,  0.0,   
             0.2, -0.4,  0.0,   
            -0.2,  0.2,  0.0,  

             0.1,  0.1,  0.0,  
             0.9,  0.1,  0.0,  
             0.5,  0.7,  0.0,   
        ], dtype="f4")

        indices = np.array([
            0, 1, 2,  
            3, 4, 5,   
            0, 1, 4,  0, 4, 3,   
            1, 2, 5,  1, 5, 4,  
            2, 0, 3,  2, 3, 5   
        ], dtype="i4")

        vbo = self.ctx.buffer(verts.tobytes())
        vba = self.ctx.buffer(indices.tobytes())

        # leagă bufferul de atributul "in_pos"
        self.vao = self.ctx.vertex_array(self.prog, [(vbo, "3f", "in_pos")], index_buffer=vba)

        self.ctx.wireframe = True

    def on_render(self, time: float, frame_time: float):
        self.ctx.clear(0.1, 0.1, 0.12)
        self.vao.render(mode=self.ctx.TRIANGLES)

    def key_event(self, key, action, modifiers):
        if key == self.wnd.keys.ESCAPE:
            self.wnd.close()

    def on_resize(self, width: int, height: int):
        self.ctx.viewport = (0, 0, width, height)

mglw.run_window_config(App, args=[])
```
